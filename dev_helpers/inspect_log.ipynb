```python
"""
Helper script to inspect a log file for specific vehicle states.
It extracts all the lines related to a specific vehicle ID. Only displays the state information for that vehicle.
"""

import ast

logname = "../models/2025-08-15_20-39-17_v0.8.0_basic_straight100Test_5000MEV_150NMEV_GREEDY/evaluation/2025-08-15_20-39-17_v0.8.0_basic_straight100Test_5000MEV_150NMEV_GREEDY_5000vehicles_evaluation.log"
target_id = "member_ev_997"
state_markers = ("State:", "Intermediate simulation state:")

def extract_after_marker(line: str):
    """Return (marker, dict_str) if a known marker is present, else (None, None)."""
    for m in state_markers:
        pos = line.find(m)
        if pos != -1:
            return m, line[pos + len(m):].strip()
    return None, None

with open(logname, "r", encoding="utf-8", errors="ignore") as f:
    for lineno, line in enumerate(f, start=1):
        if target_id not in line:
            continue  

        marker, dict_str = extract_after_marker(line)
        if marker:
            # Parse the state dict that follows the marker
            try:
                state_dict = ast.literal_eval(dict_str)
            except (SyntaxError, ValueError) as e:
                print(f"Line {lineno}: Could not parse {marker.rstrip(':')} - {e}")
                continue

            # Print only the target_id's state
            if isinstance(state_dict, dict) and target_id in state_dict:
                print(f"Line {lineno}: {state_dict[target_id]}")
            else:
                print(f"Line {lineno}: ({target_id} not present in parsed {marker.rstrip(':')})")
        else:
            # print all other lines that contain the target_id
            print(f"Line {lineno}: {line.rstrip()}")
```
