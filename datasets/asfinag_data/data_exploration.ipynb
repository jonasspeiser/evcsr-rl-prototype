# ASFINAG dataset exploration

**Source**: Autobahnen- und Schnellstraßen-Finanzierungs-Aktiengesellschaft, Österreich  
**Link**: https://www.asfinag.at/verkehr-sicherheit/verkehrszaehlung/

```python
import pandas as pd

def print_metadata(current_df):
    print(f'Das Dataframe hat {len(current_df)} Einträge')
```

```python
df = pd.read_excel("Jahr23_ASFINAG_Verkehrsstatistik.xls")
print_metadata(df)
df.head()
```

```python
df = df[df["Autobahn"].notna() & df["Autobahn"].str.startswith("A")]
print_metadata(df)
```

```python
df = df[df["Fahrzeugklasse"] == "Kfz <= 3,5t hzG"]
print_metadata(df)
```

```python
df = df[df["DTVMS"] != -1]
print_metadata(df)
df.head()
```

```python
df = df.drop_duplicates(subset=["Station"], keep="last")
print_metadata(df)
df.head()
```

```python
df = df[df["Geschätzte Tage"] <= 36]
print_metadata(df)
df.head()
```

```python
value_counts = df['Autobahn'].value_counts(dropna=False)
sorted_value_counts = value_counts.sort_index()
print(sorted_value_counts)
```

```python
pd.to_pickle(df, "auswahl-zaehlstellen_36.pickle")
```
